## The News

There was quite the gap between Singapore and COTA, but the news was dominated by one major story, a story that has been brewing for quite some time but one that I think a lot of us were surprised about.

### McLaren's Right Of Review

In the build-up to the Mexican Grand Prix, McLaren submitted a right of review concerning the penalty that Lando Norris received in the United States Grand Prix during his late-race battle with max Verstappen. If you recall, Norris was handed a five-second penalty for overtaking Max off-track as the two battled over the final podium position, demoting the Brit to fourth place. For McLaren to be able to do this, they would need to have evidence that was not available to the stewards at the time of the incident. Which, as far as I can tell, they didn't. The right of review was dismissed quickly and the penalty stood. However, this is quite an interesting tactic from McLaren, because I don't believe they intended to undo the penalty at all. In my opinion, and it is just my opinion, McLaren did this because of the nature of the Mexican circuit, and to keep the conversation over overtaking rules in the forefront of the media. There is much debate about the rules, and why they need changing, going on online, but the Woking outfit probably want to keep the discussion going and prompt change. It also helps to distract Red Bull from their weekend as they will be answering questions from the media about the process, a trick McLaren probably learnt from Mercedes in 2021.

### Alonso At 400

Fernando Alonso is the first driver in the history of Formula 1 to complete 400 Grand Prix, or at least he will be by the time he starts the race in Mexico. That's quite the achievement, since the nearest challengers to that tally are Kimi Raikkonen with 353 and Lewis Hamilton with 352. The next closest challenger, who is still racing, is Sergio Perez with 281. So it's safe to say that Alonso's record is fairly safe. With the number of races per season in the modern era of Formula 1, however, it won't be long before the likes of Max Verstappen, Charles Leclerc, and Lando Norris start cracking into the top ten total entries list. I should clarify, that while this is Alonso's 400th Grand Prix, it is only his 400th Grand Prix entry, since he has failed to start 3 of those races and only has 397 race starts. Still, that's not a bad record.

Alonso started racing in 2001, so to put that into context, there are fully grown adults, potentially with degrees, who have grown up while Alonso has been racing. It is indeed a monumental milestone across 21 seasons on the grid, but Alonso has certainly made some mistakes in his time. Fernando started his F1 career at Minardi, and if you're under the age of 25, you might not remember Minardi, but they eventually became Toro Rosso, AlphaTauri, and Racing Bulls (RB). It was, during that time, a very unimpressive car, and Alonso actually scored no points in his first season and finished 23rd in the championship, but none of that discouraged Flavio Briatore from signing Alonso to the Renault team as a test driver in 2002 and eventually as a full-time driver in 2003. The relationship eventually crumbled after the team were caught cheating, a story Alonso was instrumental in releasing, and the Spaniard moved back to Renault where he completed the 2008 season without controversy.... of course that didn't happen. The 2008 Singapore Grand Prix will be remembered forever as crash-gate. Briatore was later banned from F1. Alonso stayed at Renault until the end of 2009 when it was rumoured that he had offers from Red Bull Racing, Brawn GP (which later became Mercedes), and Ferrari. ### Race Report

Pole-sitter Lando Norris, who is famed for his lap 1 position maintaining skills, got off the line well but forgot to defend the inside line to reigning champion Max Verstappen. That sounds harsh, but in his defense, it is difficult to defend at that corner given how wide the track is. However, he can't be letting Max get up the inside like that and expect to still win the championship. The two of them arguing it out on the outside of turn 1 allowed Ferrari's Charles Leclerc to storm up the inside and take the lead of the race. From there, the Monegasque enjoyed an unchallenged drive to victory, putting on an immensely dominant display in front the the American fans. Carlos performed a tasty undercut on Norris and Verstappen to take second place, securing a Ferrari 1-2. If you haven't seen the Italian version of the commentary as those two crossed the line, I cannot recommend it enough.

Norris caught Verstappen, after performing the over-cut on his title rival, in the dying stages of the Grand Prix. What the Brit needed most was to sail on through and chase after Sainz ahead. But, as we've seen more than once with Lando, he was not able to simply overtake Max while his tyres were still good. Lap after lap Lando tried the same move down the outside of turn 12 (the hairpin at the end of the big DRS straight) with the same effect, Max kept the position. Finally, approaching the end of the Grand Prix, Lando was able to position his car on the outside and brake late into the corner. The reigning champion braked even later and ran Norris out of room on the outside. Staring down the barrel of a fourth place, and more points lost to Max, Lando booted the throttle off track and took the position anyway. This was later awarded a five-second penalty, promoting Max to third place, nearly allowing Lando's teammate Piastri to get into fourth. They finished Verstappen, Norris, Piastri, in the end.

Now, I know you're all expecting me to talk about this, so I'm not going to let my fans down. Let's discuss the overtaking first because this is becoming quite the problem in modern F1. After the turmoil that was the 2021 F1 season, the FIA changed the overtaking/defending regulation to stipulate that the car that is ahead at the apex is the defending car, and they have the right to take the racing line. This means that, as long as you are the car that is ahead, you can run your rival wide. To be considered "majority alongside", you need to have front wheels close to the other car's front wheels. How they decide that is really up to them. Now Verstappen is a master of this, and it's something we've seen from Hamilton, Schumacher, Vettel, and all multiple world champions, because that's how they become a master of the sport. So, Max pushing Lando wide at that corner was perfectly within the regulations. Now, the regulations don't state that the defending car should remain on the track, though it probably should state this, which means that when Max went off the track, there was no punishment other than a track limits strike against his name. Lando, on the other hand, overtook Max off-track. This is normally penalised with a 10-second penalty. The stewards looked at the incident and decided that because Max should have stayed on track, and didn't, then they would assign a 5-second penalty in this scenario. They are allowed to do this, of course, but I think they missed the point. They originally looked at their race position and, it seems this way, awarded a 5-second bias to Norris, rather than a 10- and 5-second to each driver, since this wouldn't affect their finishing position respective to each other. But Piastri suffered here because he did catch them and didn't have a penalty. So, what they should have done is assign the full penalties to both drivers and make them both suffer equally.

George Russell put on a masterclass of a performance to go from the pitlane to sixth position, something we are used to seeing from his teammate. After a damaged car in qualifying, George changed a heap of engine components and chose to start from the pitlane, went long in the first stint and was able to fight his way through the field in an old-spec car. Perez was behind Russell in seventh, despite starting much further up the order than the Brit, ahead of Nico Hulkenberg in eighth place, bringing home even more points for Haas. Liam Lawson went relatively under the radar in the sprint race, despite his antics with Fernando Alonso, but managed to bring his RB home in ninth place to score 2 points on his return to the sport, and beat his more-established teammate. Franco Colapinto brought his Williams home for the final point, ahead of Magnussen, Gasly, and Alonso. Tsunoda could only manage fourteenth and was furious about how he had been over-cut by his teammate over the radio, which is probably a reason he has yet to be considered for the move up to the big team. Lance Stroll managed fifteenth ahead of Albon in sixteenth in his 100th Grand Prix start.

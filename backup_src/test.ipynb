```python
import pandas as pd
import GeneralUtils.DataIO as io

from pathlib import Path

root = Path().absolute()
path_to_file = Path(root, 'Data', '2024', 'Lineup', 'Results.json')
data = io.load_json(path_to_file)
config_path = Path(root, 'config', 'SeasonInfo.json')
info = io.load_json(config_path)
races = info['2024']['Races']
```

```python
def create_dataframe(race, races, dictionary):
    race_index = races.index(race)
    number_of_races = race_index + 1
    races_so_far = races[0: number_of_races]
    data_dict = {}
    for key, values in dictionary.items():
        data_dict.update({key: values[0: number_of_races]})
    df = pd.DataFrame(data_dict)
    df.index = races_so_far
    styled_df = df.style.background_gradient(cmap='RdYlGn')
    return styled_df
```

```python
dic = data['Driver Points']
race = 'Brazil'
table = create_dataframe(race, races, dic)
table
```

```python
dic = data['Driver Values']
table = create_dataframe(race, races, dic)
table
```
